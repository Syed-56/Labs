```python
import socket
```

1. Getting IP Address of host

```python
hostname = socket.gethostname()
ip_address = socket.gethostbyname(hostname)
print("Hostname = ", hostname)
print("IP Address = ", ip_address)
```

2. Getting IP Addresses of any host

```python
hostname = ['google.com','treknuces.app']
for i in hostname:
    ip_address = socket.gethostbyname(i)
    print("Hostname = ", i)
    print("IP Address = ", ip_address)
```

3. Getting hostname by IP address

```python
hostname1 = socket.gethostbyaddr('192.168.1.10')
hostname2 = socket.gethostbyaddr('8.8.8.8')
print("IP Address - Host 1 = ", hostname1)
print("IP Address - Host 2 = ", hostname2)
```

4. Getting Service name, Given port number, Protocol

```python
def find_service_name():
    protocol_name = 'tcp'
    for port in [80,25]:
        servicename_tcp = socket.getservbyport(port,protocol_name)
        print(f"Port {port} -> Service Name: {servicename_tcp}")
    servicename_udp = socket.getservbyport(53,'udp')
    print(f"Port {53} -> Service Name: {servicename_udp}")

find_service_name()
```

5. Port Scanner

```python
from socket import *
import time

startTime = time.time()

target = input("Enter Hostname to be Scanned: ")
targetIP = gethostbyname(target)
print("Scanning Host ", targetIP)

for i in range(50,500):
    s = socket(AF_INET, SOCK_STREAM)
    conn = s.connect_ex((targetIP,i))
    
    if (conn == 0):
        print(f"Port {i}: OPEN")
    s.close()

timeTaken = time.time()-startTime
print("Time Taken in Seconds = ", timeTaken)
print("Time Taken in Minutes = ", time.time()-startTime/60)
```

6. Client Server Connection Program
- Create 2 different files of client and server and run them to see in action

```python
#Server
s = socket.socket()
print("Socket Created")
s.bind(('localhost',9999))    #binding to port
s.listen(5)     #wait for client
print("waiting for connection")

while True:
    c, addr = s.accept()    #establish connection with client
    print("Got Connection from ", addr)
    c.send(bytes("Thank You For Connecting"))
    c.close()
```

```python
#client
s = socket.socket()
print("Socket Created")
s.bind(('localhost',9999))
print(s.recv(1024).decode())
s.close()
```
